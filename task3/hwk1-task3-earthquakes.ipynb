# 6
Without creating any new variables, plot a histogram of the eartquake's magnitudes (`eqk` data). Add a title and change the color of your graph.

```python
# Plotting the earthquake's magnitude as a histogram
eqk['mag'].plot.hist(                                  # Specify data frame with specific column
                    bins = 60,                         # Using rice rule for amount of bins
                    color = 'red',                     # Setting the color
                    edgecolor = 'black',               # Adding a border to the columns
                    title = "Earthquake Magnitudes",   # Adding representative title
                    xlabel = "Magnitude",              # Adding x-axis title
                    ylabel = "Count")                  # Adding y-axis title

# Display the plot
plt.show()
```

# 7

a. Create a new variable `top20` with the 20 earthquakes with largest magnitude. HINT: check the method [`pandas.Series.nlargest`](https://pandas.pydata.org/docs/reference/api/pandas.Series.nlargest.html). Your answer should contain only the `id` and `mag` columns. 

```python
# Saving the 20 largest magnitude earthquakes
top20 = eqk['mag'].nlargest(20)
```

```python
# View the dataframe 
top20
```

```python
grader.check("q7_a")
```

<!-- BEGIN QUESTION -->

b. Create a bar plot showing the magnitude of the top 20 earthquakes. Update the title, x-axis label, y-axis label, and color of the bars. 

```python
# Create bar plot of the top 20 earthquake magnitudes
top20.plot.bar(
            top20.index, top20,                          # Specify data
            color = 'red',                               # Setting the color
            edgecolor = 'black',                         # Adding a border to the columns
            title = "Top 20 Earthquakes by Magnitude",   # Adding a representative title
            xlabel = "Earthquake ID",                    # Adding x-axis title
            ylabel = "Magnitude",                        # Adding y-axis title
            rot = 80)                                    # Rotating columns

# Display the plot
plt.show()
```

<!-- END QUESTION -->

<!-- BEGIN QUESTION -->

c. What is the role of the `top20` index in the graph? What would have happened if we had not updated the index?

> The role of the `top20` index is to label each bar on the x-axis with the earthquake's unique ID, making it clear which earthquakes have the largest magnitudes. If we had not set the index to the earthquake `id`, the index would still be the original row number from the entire dataset. Since the top 20 earthquakes are scattered throughout different rows, the bars would be plotted at those row positions along a  x-axis, making them extremely small and far apart.

# 9 

Without creating any new variables, visualize the locations of eartquakes with magnitude greter than 5 by creating a scatter plot of their latitude and longitude. Make the following adjustments to the basic plot:
- update the title
- color the points by magnitude (HINT: [check the `c` parameter here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.plot.scatter.html))
- adjust the `alpha` (transparency) argument
- adjust the `colormap` argument to maket it look prettier. [Click for ideas of sequential colormaps.](https://matplotlib.org/stable/users/explain/colors/colormaps.html)


```python
# Geographically plotting Earthquakes with magnitude greater than 5
eqk[eqk['mag'] > 5].plot.scatter(
    x = 'longitude',                                        # Add x-axis title
    y = 'latitude',                                         # Add y-axis title
    c = 'mag',                                              # Color points by magnitude
    alpha = 0.3,                                            # Make points more transparent
    colormap = 'viridis',                                   # Use sequential color map
    title = "Earthquakes with a Magnitude Greater than 5")  # Adding a representative title

# Display the plot
plt.show()
```
